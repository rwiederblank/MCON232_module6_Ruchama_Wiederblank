```python
def load_words(filename: str):
    words = []
    with open(filename, "r") as f:
        for line in f:
            line = line.strip()
            words.append(line)
    return words

def normalize(text : str):
    stripped = ""
    for char in text:
        if char.isalpha:
            stripped += char.lower()
    return stripped

def set_signature(text : str):
    text = normalize(text)
    fs = frozenset(text)
    return fs

def freq_signature(text : str):
    text = normalize(text)
    my_lst = "abcdefghijlkmnopqrstuvwxyz"
    freq = []
    for char in my_lst:
        if char not in text:
            freq.append(0)
        else:
            count = 0
            for letter in text:
                if letter == char:
                    count += 1
            freq.append(count)
    freq = tuple(freq)
    return freq

def is_anagram_using_sets(a : str, b : str):
    return set_signature(a) == set_signature(b)

def is_anagram(a : str, b : str):
    return freq_signature(a) == freq_signature(b)

def group_anagrams(words : list[str]):
    d = {}
    for word in words:
        if freq_signature(word) not in d:
            d[freq_signature(word)] = [word]
        else:
            d[freq_signature(word)].append(word)
    return d

# running and printing:
words = load_words("words.txt")
d = group_anagrams(words)
groups = list(d.values())
singltons = 0
anagrams = 0
largest = 0
largest_key = ""
lengths = []
for key, value in d.items():
    lengths.append(value)
    if len(value) == 1:
        singltons += 1
    else:
        anagrams += 1
    if len(value) > largest:
        largest = len(value)
        largest_key = key

# set counterexamples
print("SET COUNTEREXAMPLES (set method says True, but NOT anagrams): ")
print("aab vs ab: ", is_anagram_using_sets("aab", "ab"))
print("mississippi vs misp: ", is_anagram_using_sets("mississippi", "misp"))
print("adder vs read: ", is_anagram_using_sets("adder", "read"))
print()
print("ANAGRAM GROUP REPORT ")
print(f"Total words loaded: {len(words)}")
print(f"Total anagram groups: {len(d)}")
print(f"Singleton groups (size 1): {singltons}")
print(f"Largest group (size {largest}):")
print(d[largest_key])
print()
print("TOP 5 GROUPS: ")
top5 = []
for i in range(5):
    longest = max(lengths, key=len)
    print(f"Group size {len(longest)}: {longest}")
    lengths.remove(longest)

# short answers
print()
print("SHORT ANSWER QUESTIONS")
print("Question 1: x in my_set is average-case O(1) because sets use hashing to locate what it needs, so it does not have to go through the whole set, instead can go exactly to the place it needs.")
print("Question 2: x in my_list is average-case O(n) because it needs to loop through the array and check every value until it finds the correct one. ")
print("Question 3: Building a frequency is 0(k) because it needs to go through every single element of the word. ")
print("Question 4: Sets are not good for anagrams because they get rid of duplicates, so it may say some words are anagrams if they have the same letters, when they are not. ")

```
